# 15. Correlações customizadas no ydata-profiling

O `ydata-profiling` permite configurar diferentes métodos de correlação.

Alguns métodos comuns:

- **Pearson**: relação linear entre variáveis numéricas;
- **Spearman**: relação monotônica, mesmo que não seja perfeitamente linear;
- **Kendall**: útil para relações ordinais;
- **Cramér's V**: associação entre variáveis categóricas;
- **Phi K**: associação entre variáveis de tipos mistos.

Importante:

> Correlação ou associação não significa causalidade.

````python
profile_correlacoes = ProfileReport(
    df,
    title="Relatório com correlações customizadas",
    explorative=True,
    correlations={
        "auto": {
            "calculate": True,
            "warn_high_correlations": True,
            "threshold": 0.8
        },
        "pearson": {
            "calculate": True,
            "warn_high_correlations": True,
            "threshold": 0.8
        },
        "spearman": {
            "calculate": True,
            "warn_high_correlations": True,
            "threshold": 0.8
        },
        "kendall": {
            "calculate": False
        },
        "phi_k": {
            "calculate": True,
            "warn_high_correlations": True,
            "threshold": 0.8
        },
        "cramers": {
            "calculate": True,
            "warn_high_correlations": True,
            "threshold": 0.8
        }
    }
)

profile_correlacoes.to_file("relatorio_correlacoes_customizadas.html")
````

# 17. Criação de uma variável de data

A base possui ano, mês e dia da chegada separados.  
Podemos criar uma coluna de data para analisar sazonalidade.

Atenção: a coluna de mês está em texto, então precisamos combinar corretamente.

````python
# Verifique os nomes dos meses
df["arrival_date_month"].unique()
````

````python
# Criando uma coluna de data de chegada

df["arrival_date"] = pd.to_datetime(
    df["arrival_date_year"].astype(str) + "-" +
    df["arrival_date_month"].astype(str) + "-" +
    df["arrival_date_day_of_month"].astype(str),
    errors="coerce"
)

df[["arrival_date_year", "arrival_date_month", "arrival_date_day_of_month", "arrival_date"]].head()
````

````python
df["arrival_date"].isna().sum()
````
